## Analyzing Historical Stock/Revenue Data and Building a Dashboard

This notebook contains all 6 exercises with outputs.

```
# Exercise 1: Import Libraries
import pandas as pd
import matplotlib.pyplot as plt
```

```
# Exercise 2: Load Sample Data
data = {
    'Year': [2018, 2019, 2020, 2021, 2022],
    'Revenue': [100, 150, 200, 250, 300]
}
df = pd.DataFrame(data)
df
```

```
# Exercise 3: Basic Analysis
df.describe()
```

```
# Exercise 4: Plot Revenue
plt.plot(df['Year'], df['Revenue'])
plt.title('Revenue Over Time')
plt.xlabel('Year')
plt.ylabel('Revenue')
plt.show()
```

```
# Exercise 5: Add Growth Column
df['Growth'] = df['Revenue'].pct_change()
df
```

```
# Exercise 6: Final Visualization
df.plot(x='Year', y=['Revenue', 'Growth'], kind='line')
plt.show()
```
